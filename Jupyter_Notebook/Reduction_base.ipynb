# Problème du vecteur le plus proche dans une base orthogonale

**Proposition**  
Soit $B=(b_1,\dots,b_n)$ une base **orthogonale** de $\mathbb{R}^n$, c’est‑à‑dire  
$$
\langle b_i,b_j\rangle = 0 \quad(i\neq j),
\qquad
\|b_i\|_2>0.
$$
Écrivons  
$$
v = \sum_{i=1}^n x_i\,b_i,
\quad x_i\in\mathbb{R},
$$
et posons  
$$
\Lambda
= (B)
= \bigl\{\,w = \sum_{i=1}^n m_i\,b_i \mid m_i\in\mathbb{Z}\bigr\}.
$$
Alors un vecteur $w^*\in\Lambda$ qui minimise la distance euclidienne $\|v - w\|_2$ est

$$
w^* = \sum_{i=1}^n \lfloor x_i\rceil\,b_i,
$$

où $\lfloor x_i\rceil$ désigne un entier **le plus proche** de $x_i$.

---

## Preuve

1. **Décomposition de la distance**  
   Pour tout  
   $$
   w = \sum_{i=1}^n m_i\,b_i \in \Lambda,
   $$
   on a  
   $$
   v - w
   = \sum_{i=1}^n (x_i - m_i)\,b_i,
   $$
   et, par orthogonalité des $b_i$,  
   $$
   \|v - w\|_2^2
   = \Bigl\|\sum_{i=1}^n (x_i - m_i)\,b_i\Bigr\|_2^2
   = \sum_{i=1}^n (x_i - m_i)^2\,\|b_i\|_2^2.
   $$

2. **Minimisation coordonnée**  
   Minimiser $\|v - w\|_2$ revient à minimiser  
   $\sum_{i=1}^n (x_i - m_i)^2\,\|b_i\|_2^2$.  
   Chaque terme ne dépend que de $m_i$, et comme $\|b_i\|_2^2>0$,  
   $$
   m_i^*
   = \arg\min_{m\in\mathbb{Z}} (x_i - m)^2
   = \lfloor x_i\rceil.
   $$

3. **Propriété d’arrondi**  
   Pour tout réel $x$ et son entier le plus proche $m=\lfloor x\rceil$,  
   $\lvert x - m\rvert \le \tfrac12$.  
   Chaque $(x_i - m_i^*)^2$ est ainsi minimal.

4. **Conclusion**  
   Le vecteur
   $$
   w^* = \sum_{i=1}^n m_i^*\,b_i,
   \quad m_i^* = \lfloor x_i\rceil,
   $$
   réalise la distance minimale $\|v - w\|_2$ parmi tous les $w\in\Lambda$.


### Fonction `gener_base(n, d)`

**Description**  
Génère une base $B \in \mathbb{Z}^{n \times n}$ dont les coefficients sont choisis uniformément  
dans l’intervalle $[-d, d]$, et garantit que $B$ est inversible (car base).

**Paramètres**  
- **`n`** (_int_)  
  Dimension de l’espace (nombre de vecteurs, taille de la base).  
- **`d`** (_int_)  
  Valeur absolue maximale pour la génération aléatoire des coefficients (entiers dans $[-d,d]$).

**Retour**  
- **`B`** (_ndarray_ de forme `(n, n)`)  
  Matrice inversible représentant la base générée.



```python
import numpy as np

def gener_base(n, d):
    while True:
        B = np.random.randint(-d, d+1, size=(n, n))
        if np.linalg.matrix_rank(B) == n:
            return B


B = gener_base(4, 5)
print("Base générée B =\n", B)
```

**Fonction** `gener_unimodulaire(n, a, b)`

Génère une matrice unimodulaire $U \in \mathbb{Z}^{n\times n}$ en composant a transformations unimodulaires élémentaires choisies aléatoirement :
- **Permutation** de deux colonnes  
- **Changement de signe** d’une colonne  
- **Addition** d’un multiple entier $k \in [-b, b]$ d’une colonne à une autre  

**Paramètres**  
- `n` (_int_) : dimension de la matrice \((n\times n)\).  
- `a` (_int_) : nombre de transformations élémentaires à composer.  
- `b` (_int_) : valeur maximale absolue pour le multiple \(k\) dans les additions de colonnes.  

**Retour**  
- `U` (_ndarray_ de forme `(n, n)`) : matrice unimodulaire obtenue (determinant = ±1).  


```python
import numpy as np

def gener_unimodulaire(n, a, b):

    U = np.eye(n, dtype=int)
    for _ in range(a):
        op = np.random.choice(['permute', 'sign', 'add'])
        if op == 'permute':
            i, j = np.random.choice(n, size=2, replace=False)
            # échange des colonnes i et j
            U[:, [i, j]] = U[:, [j, i]]
        elif op == 'sign':
            i = np.random.randint(0, n)
            # changement de signe de la colonne i
            U[:, i] *= -1
        else:  # add
            src, dst = np.random.choice(n, size=2, replace=False)
            k = np.random.randint(-b, b+1)
            # addition de k fois la colonne src à la colonne dst
            U[:, dst] += k * U[:, src]
    return U

# Exemple d'utilisation
U = gener_unimodulaire(5, 10, 3)
U
```

### Fonction `gener_vect(n, b)`

**Description**  
Génère un vecteur v dans Z^n dont les coefficients sont choisis uniformément  
dans l’intervalle [-b, b].

**Paramètres**  
- `n` (int) : taille du vecteur.  
- `b` (int) : valeur absolue maximale des coefficients (entiers dans [-b, b]).

**Retour**  
- `v` (ndarray de forme (n,)) : vecteur aléatoire généré.



```python
import numpy as np

def gener_vect(n, b):
    v = np.random.randint(-b, b + 1, size=(n,))
    return v

# Exemple d'utilisation
v = gener_vect(5, 10)
print("Vecteur v =", v)
```

# Réduction LLL d’une base entière

Soit $B\in\mathbb{Z}^{n\times n}$ une base de $\mathbb{R}^n$. On note $(B^*,M)$ sa décomposition de Gram–Schmidt,  
c’est‑à‑dire  
$$
B = B^*\,M,
$$  
où $B^*$ a des colonnes orthogonales et $M$ est triangulaire supérieure à 1 sur la diagonale.

On définit ensuite :

1. **normalize** : pour chaque $j>i$, on arrondit $\mu_{i,j}$ à l’entier $k=\lfloor\mu_{i,j}\rceil$ puis on fait  
   $$
   B[:,j]\;-\!=\;k\,B[:,i],\qquad M_{i,j}\;-\!=\;k.
   $$
2. **condition** : pour $0\le k<n-1$ et , on vérifie  
   $$
   3/4\times|b^*_k\|^2 \;\le\;\|\,b^*_{k+1} + \mu_{k,k+1}\,b^*_k\|^2.
   $$
3. **GSO_opt** : si la condition échoue en $k$, on échange les colonnes $k$ et $k+1$ de $B$,  
   puis on met à jour localement $B^*[:,k]$, $B^*[:,k+1]$ et les coefficients  
   $\mu_{i,k},\mu_{i,k+1}$ pour $i\le k+1$.


```python
import numpy as np

def GSO(B):
    """Retourne (Bo, M) pour B = Bo @ M avec décomp. de Gram–Schmidt."""
    n = B.shape[1]
    Bo = np.zeros_like(B, dtype=float)
    M = np.eye(n, dtype=float)
    for j in range(n):
        v = B[:, j].astype(float)
        for i in range(j):
            mu = (Bo[:, i] @ B[:, j]) / (Bo[:, i] @ Bo[:, i])
            M[i, j] = mu
            v -= mu * Bo[:, i]
        Bo[:, j] = v
    return Bo, M

def normalize(B, M):
    """Normalise M via opérations unimodulaires sur B en parallèle."""
    n = M.shape[0]
    for j in range(n):
        for i in range(j):
            kij = int(round(M[i, j]))
            if kij != 0:
                # Opération unimodulaire : C_j -= kij * C_i
                B[:, j] -= kij * B[:, i]
                M[i, j] -= kij
    # Ne modifie pas Bo

def condition(Bo, M, k, delta=0.75):
    """Teste la condition LLL pour l’indice k."""
    b_k_norm2 = Bo[:, k] @ Bo[:, k]
    proj = Bo[:, k+1] + M[k, k+1] * Bo[:, k]
    return delta * b_k_norm2 <= (proj @ proj)

def GSO_opt(B, Bo, M, j):
    """Échange colonnes j et j+1 de B puis met à jour Bo et M localement."""
    # 1) échange de colonnes j <-> j+1
    B[:, [j, j+1]] = B[:, [j+1, j]]
    # 2) recalcul partiel de GSO pour colonnes j et j+1
    n = B.shape[1]
    for col in (j, j+1):
        v = B[:, col].astype(float)
        for i in range(col):
            mu = (Bo[:, i] @ B[:, col]) / (Bo[:, i] @ Bo[:, i])
            M[i, col] = mu
            v -= mu * Bo[:, i]
        Bo[:, col] = v
    # 3) on doit mettre à jour M[k,j+1] pour k < j
    #    mais l’opération ci‑dessus l’a déjà fait pour j et j+1

def LLL_reduce(B, delta=0.75):
    """Réduit la base B selon l’algorithme LLL en place."""
    Bo, M = GSO(B)
    n = B.shape[1]
    k = 0
    while k < n - 1:
        normalize(B, M)
        if condition(Bo, M, k, delta):
            k += 1
        else:
            GSO_opt(B, Bo, M, k)
            k = max(k-1, 0)
    return B
```

```python
import numpy as np
import pandas as pd

def experiment(n, d, a, b):
    # 1. génération
    B    = gener_base(n, d)
    U    = gener_unimodulaire(n, a, b)
    v    = gener_vect(n, b)

    # 2. constitution des 4 variantes de base
    B1 = B
    B2 = LLL_reduce(B.copy())
    B3 = B @ U
    B4 = LLL_reduce(B3.copy())

    # 3. calcul inline des distances
    def dist(Bi):
        xi = np.linalg.solve(Bi, v)
        mi = np.rint(xi).astype(int)
        wi = Bi @ mi
        return np.linalg.norm(v - wi)

    distances = [dist(B1), dist(B2), dist(B3), dist(B4)]

    # 4. mise en forme
    df = pd.DataFrame({
        'Base': [
            'B original',
            'B LLL réduite',
            'B·U',
            'B·U LLL réduite'
        ],
        'Distance': distances
    })
    return B, U, v, df
```

```python
# Exemple 1 : paramètres de base
B, U, v, df = experiment(n=4, d=5,  a=10, b=3)
print("Exemple 1 – v =", v)
print(df)

# Exemple 2 : base un peu plus large, plus d’opérations unimodulaires
B, U, v, df = experiment(n=4, d=8,  a=15, b=4)
print("Exemple 2 – v =", v)
print(df)

# Exemple 3 : dimension supérieure
B, U, v, df = experiment(n=5, d=5,  a=10, b=3)
print("Exemple 3 – v =", v)
print(df)

# Exemple 4 : plus de transformations, vecteur plus grand
B, U, v, df = experiment(n=4, d=5,  a=20, b=6)
print("Exemple 4 – v =", v)
print(df)

# Exemple 5 : très large coefficients
B, U, v, df = experiment(n=4, d=20, a=10, b=10)
print("Exemple 5 – v =", v)
print(df)
```

## Conclusion des expériences

1. **Transformation unimodulaire sans réduction**  
   Appliquer $B \mapsto B \cdot U$ sans réduction LLL fait souvent passer la distance d’approximation  
   de quelques unités à des dizaines, voire des centaines. Le réseau reste le même, mais la base devient  
   très « tordue », et l’arrondi devient catastrophique.

2. **Réduction LLL sur la base originale**  
   - Dans la plupart des cas (ex. 1, 3, 4, 5), LLL fait **chuter** la distance d’approximation (parfois très nettement, ex. 3 et 5).  
   - Il peut exister des cas marginaux (ex. 2) où l’erreur augmente légèrement, mais globalement LLL **tend à orthogonaliser** la base et à réduire le défaut.

3. **Réduction LLL après transformation**  
   Même lorsqu’une transformation $U$ fait exploser la distance, l’application de LLL **ramène**  
   systématiquement l’erreur à un niveau comparable (ou identique) à celui de la base originale réduite.

---

### Recommandation

**Toujours** appliquer la réduction LLL avant tout calcul d’approximation par arrondi dans une base entière :  
- Elle garantit une **qualité** d’approximation élevée.  
- Elle est **robuste** face à toute transformation unimodulaire.  
- Sans LLL, l’erreur peut devenir très importante.
